```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]
plt.plot(x, y)
plt.show()
```

Recommended Approach:

```python
# An axes is a subplot
fig, ax = plt.subplots()
ax.plot(x, y)   # for line plots
ax.set(title="Simple Plot",
       xlabel="x-axis",
       ylabel="y-axis");
```

# Plot Types

## Scatter Plot

```python
x = np.linspace(0, 10, 100)
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x));
```

## Bar Plot

```python
nut_butter_prices = {
    "Almond": 10,
    "Peanut": 8,
    "Cashew": 12
}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values());
```

## Histogram

```python
# Create 1000 random points within the standard normal distribution
x = np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(x);
```

# Multiple Plots

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2,
                                             ncols=2,
                                             figsize=(10, 5))
ax1.plot(x, x/2);
ax2.scatter(np.random.random(10), np.random.rand(10));
ax3.barh(list(nut_butter_prices.keys()), list(nut_butter_prices.values()));
ax4.hist(np.random.randn(1000));
```

# Using Pandas DataFrames

```python
car_sales = pd.read_csv("https://raw.githubusercontent.com/mrdbourke/zero-to-mastery-ml/refs/heads/master/data/car-sales.csv")
car_sales
```

```python
car_sales['Price'] = car_sales['Price'].str.replace('[\$\,]', '', regex=True).astype(float)
car_sales
```

```python
car_sales["Sale Date"] = pd.date_range("1/1/2020", periods=len(car_sales))
car_sales
```

```python
car_sales["Total Sales"] = car_sales["Price"].cumsum()
car_sales
```

```python
# Call .plot() directly on the dataframe
# wrapper for matplotlib's plot()
car_sales.plot(x="Sale Date", y="Total Sales");
```

```python
car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter");
```

# Heart Disease Example

```python
heart_disease = pd.read_csv("heart-disease.csv")
heart_disease.head()
```

```python
heart_disease["age"].plot.hist();
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
print(f"There are {len(over_50)} observations in this dataset")
```

Pyplot method

```python
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target");
```

OO and pyplot methods combined

```python
fig, ax = plt.subplots()
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax);
```

OO method

```python
fig, ax = plt.subplots()

scatter = ax.scatter(x=over_50["age"],
                     y=over_50["chol"],
                     c=over_50["target"])

ax.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol")

ax.axhline(over_50["chol"].mean(),
           linestyle="--")

ax.legend(*scatter.legend_elements(), title="Target");
```

```python
fig, (ax0, ax1) = plt.subplots(figsize=(10, 10),
                               nrows=2,
                               ncols=1,
                               sharex=True)

# For chol vs age
scatter = ax0.scatter(x=over_50["age"],
                     y=over_50["chol"],
                     c=over_50["target"])

ax0.set(title="Heart Disease and Cholesterol Levels",
       ylabel="Cholesterol")

ax0.axhline(over_50["chol"].mean(),
           linestyle="--");

ax0.legend(*scatter.legend_elements(), title="Target");

# For thalach vs age
scatter = ax1.scatter(x=over_50["age"],
                     y=over_50["thalach"],
                     c=over_50["target"])

ax1.set(title="Heart Disease and Max Heart Rate",
       xlabel="Age",
       ylabel="Max Heart Rate")

ax1.axhline(over_50["thalach"].mean(),
           linestyle="--")

ax1.legend(*scatter.legend_elements(), title="Target");

# Figure title
fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");
```
